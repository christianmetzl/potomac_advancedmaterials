```python
import subprocess, sys, os, tempfile
ROOT = os.getcwd()
runs = [
    ('src/stage2_refinement.py',           'Two-stage GQE (H2/H4/H6) -> chemical accuracy (0.000/0.009/0.297 mHa)'),
    ('src/transition_metal_oxide_qsci.py', 'CrO 5-Pi / NiO 3-Sigma 20q multireference vs exact CASCI (0.038/0.197 mHa)'),
    ('src/sno_demo.py',                     'SnO 16q EUV target, chemical accuracy'),
    ('src/hamlib_validate.py 14',           'HamLib third-party Hamiltonian equivalence @28q (term count + one-norm)'),
]
with tempfile.TemporaryDirectory() as wd:   # keep the cloned workspace pristine; scratch files land here
    for script, note in runs:
        parts = script.split(); parts[0] = os.path.join(ROOT, parts[0])
        print('='*78); print('>>>', note); print('    $ python', script, flush=True)
        r = subprocess.run([sys.executable, *parts], cwd=wd, capture_output=True, text=True)
        out = (r.stdout or '') + (r.stderr or '')
        print('\n'.join(out.strip().splitlines()[-14:]))
print('='*78); print('Fast verification complete.')
```
